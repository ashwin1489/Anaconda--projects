```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years, runs)
plt.show()
```

```python
plt.plot(years, runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
plt.show()
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years, kohli, label="Virat Kohli")
plt.plot(years, sehwag, label="Virender Sehwag")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance Comparison")
plt.legend()
plt.show()
```

```python
plt.plot(years, kohli, 'ro--', label="Kohli")  # red circles with dashed lines
plt.plot(years, sehwag, 'g^:', label="Sehwag")  # green triangles dotted
plt.legend()
```

```python
with plt.xkcd():
    plt.plot(years, kohli, label="Kohli")
    plt.plot(years, sehwag, label="Sehwag")
    plt.title("Epic Battle of the Batsmen")
    plt.legend()
    plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Data preparation
years = list(range(2010, 2022))  # X-axis (10 years)

# Hypothetical runs for each player (Y-axis values)
kohli_runs = [800, 900, 1100, 1200, 1300, 1250, 1400, 1500, 1600, 1700,1800,1900]
rohit_runs = [700, 800, 950, 1100, 1200, 1150, 1300, 1450, 1500, 1550,1600,1700]
sehwag_runs = [1000, 1200, 1150, 1000, 900, 800, 750, 700, 650, 600,700,800]

# 2. Plotting each player's data
plt.plot(years, kohli_runs, label="Virat Kohli", color="red", linestyle="-", linewidth=2)
plt.plot(years, rohit_runs, label="Rohit Sharma", color="blue", linestyle="--", linewidth=2)
plt.plot(years, sehwag_runs, label="Virender Sehwag", color="green", linestyle=":", linewidth=2)

# 3. Custom style (markers for Kohli’s line)
plt.plot(years, kohli_runs, marker="o", color="red", linestyle="-", linewidth=2)

# 4. Adding labels and title
plt.xlabel("Year")                # X-axis label
plt.ylabel("Runs Scored")         # Y-axis label
plt.title("Runs Comparison of Kohli, Rohit & Sehwag (2010-2019)")  # Title of the chart

# 5. Adding legend
plt.legend()  

# 6. Adding grid for readability
plt.grid(True, linestyle="--", alpha=0.6)

# 7. Show the plot
plt.show()
```
